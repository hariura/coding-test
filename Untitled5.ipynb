```python
import sys
sys.setrecursionlimit(100000)

def solution(maps):
    answer = []
    visited = [[0] * len(maps[0]) for _ in range(len(maps))]
    dx = [-1,1, 0,0]
    dy = [0,0, -1,1]
    # curr_sum = 0 
    
    def dfs(x, y):
        curr_sum = 0
        # 종료조건
        if x < 0 or y < 0 or x>= len(maps) or y >= len(maps[0]):
            return curr_sum
            
        if  maps[x][y] != "X" and visited[x][y] == 0 :
            visited[x][y] = 1
            
            for i in range(4):
                nx, ny = x+ dx[i], y+dy[i]
                curr_sum += dfs(nx, ny)
            return int(maps[nx][ny]) + curr_sum
        return curr_sum
    
    for i in range(len(maps)):
        for j in range(len(maps[0])):
            if  maps[i][j] != "X" and visited[i][j] == 0 :
                answer.append(dfs(i, j))
            
    return answer
```

```python
solution(maps)
```
