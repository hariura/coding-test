```python
def solution(prices):    
    answer = [0] * len(prices)
    l = range(len(prices))
    for i in l:
        for k in l[i+1:]:
            answer[i] += 1
            if prices[i] > prices[k]:
                break
    return answer
```

- 확실히 코딩테스트 합격자되기가 그림이 직관적이라 이해하기 편하다

```python
def solution(prices):
    n = len(prices)
    answer = [0] * n
    stack = [0]
    
    for i in range(len(prices)):
        #주식가격이 떨어졌어. #한번만 빠지는게 아니니까 while로 걸음 
        while stack and prices[i] < prices[stack[-1]]: 
            j = stack.pop()
            answer[j] = i-j
        stack.append(i)
    
    while stack:
        j = stack.pop()
        answer[j] = n - 1 - j 

    return answer
```
