routes.get(t[0], [])는 t[0]라는 키가 routes 딕셔너리에 존재하면 그 값을 반환하고, 존재하지 않으면 빈 리스트 []를 반환합니다.

```python
# tickets = [["ICN", "JFK"], ["HND", "IAD"], ["JFK", "HND"]]
tickets = [["ICN", "SFO"], ["ICN", "ATL"], ["SFO", "ATL"], ["ATL", "ICN"], ["ATL","SFO"]]
```

```python
routes = {}
routes
```

```python
# 이렇게 하면 없는 키라서 오류남 
# t = tickets[0]
# routes[t[0]] = routes[t[0]] + [t[1]]
# routes
```

```python
# 이렇게 넣으면 다음 벨류 넣을때 값이 겹침 
# t = tickets[0]
# routes[t[0]] = [t[1]]
# routes
```

```python
# 첫번째 넣고 두번째 넣을때 이렇게 일일이 해주는게 어려움 
# t = tickets[1]
# routes[t[0]] = routes[t[0]] + [t[1]]
# routes
```

```python
t = tickets[0]
routes[t[0]] = routes.get(t[0], []) + [t[1]]
routes
```

```python
t = tickets[1]
routes[t[0]] = routes.get(t[0], []) + [t[1]]
routes
```

```python
routes = {}
for t in tickets:
    routes[t[0]] = routes.get(t[0], []) + [t[1]]
```

```python
for r in routes:
    routes[r].sort(reverse=True)
```

```python
routes
```

```python
stack = ["ICN"]
path = []
while len(stack) > 0:
    top = stack[-1]
    if top not in routes or len(routes[top]) == 0:
        path.append(stack.pop())
    else:
        stack.append(routes[top][-1]) #스택에 쌓고
        routes[top] = routes[top][:-1] #그걸 뺀거야 
# return path[::-1]
```

```python
routes["ATL"]
```

```python
routes["ATL"] = routes["ATL"][:-1]
```

```python
tickets = [["ICN", "SFO"], ["ICN", "ATL"], ["SFO", "ATL"], ["ATL", "ICN"], ["ATL","SFO"]]
```

```python
r = defaultdict(list)
```

```python
#defaultdict를 써서 좋은점 : dict 구조를 그대로 가져가는데, 오류가 안뜸 
r = defaultdict(list)
for i,j in tickets:
    r[i].append(j)
```

```python
students = [
    ('John', 'A', 15),
    ('Jane', 'B', 12),
    ('Dave', 'B', 10),
    ('Kavin', 'B', 11)
]

# 성적으로 정렬하고, 같은 성적이면 나이가 많은 순서로 정렬
sorted_students = sorted(students, key=lambda x: (x[1], -x[2]))

print(sorted_students)
```
