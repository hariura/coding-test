```python
# student 를 studnet로 적는오타 두번이나 나왔는데 찾질 못했어. 
def solution(N, students, calls):
    answer = 0 
    for who in calls:
        index = students.index(who)
        answer += index
        
        temp = []
        for i in students:
            if i != who:
                temp.append(i)
        students = temp
    return answer 
```
