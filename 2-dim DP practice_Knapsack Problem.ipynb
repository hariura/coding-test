Knapsack Problem
- 문제 설정
- 입력: 항목들의 집합, 각 항목의 가치와 무게, 그리고 배낭의 최대 용량.
- 목표: 배낭의 용량을 초과하지 않으면서 가치의 총합을 최대화하는 항목들의 조합을 찾는 것.



항목들: A, B, C (각각의 가치와 무게가 있음)  
가치: A = 60, B = 100, C = 120  
무게: A = 10, B = 20, C = 30  
배낭 최대 용량: 50  


```python
def solution(values, weights, W):
    dp = [[0]*(W+1) for _ in range(len(values)+1)] #가치에 대한 table row는 value, col은 가능한웨이트
    for i in range(1,len(values)+1): #행
        for j in range(1,len(dp[0])): #열 
            if weights[i-1] <= j:
                dp[i][j] = max(dp[i-1][j],dp[i-1][j-weights[i-1]] +values[i-1])
            else:
                dp[i][j] = dp[i-1][j]
    print(len(dp),len(dp[0]))
    return dp[len(values)][W]
```
