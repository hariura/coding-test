```python
from collections import deque

def solution(skill, skill_trees): #순서에 맞게 잖아
    answer = 0
    
    for skills in skill_trees: #이중 몇개가 스킬에 맞는지 
        queue = deque(skill[:])
        for s in skills:
            if s in skill and s != queue.popleft(): break
            
        else : answer += 1

    
    return answer
```
