# 3월4일 실패 
점화식 한번에 세워야한대 
목표와 처음시작할 상태, 종료조건 정의 중요 

```python
def find(data, p, step):
    if len(p) == 6 : return  # step 리턴 선언하고 아무것도 안받음 
    if p != '' : data.append(p) # 순번 세려고 하는구나 
    
    for c in ['A','E','I','O','U']:
        find(data, ''.join([p,c]), step + 1 )

def solution(word):
    answer = 0
    data = []
    find(data, "",0)
    for i in range(len(data)):
        if data[i] == word:
            answer = i+1
            break
    
    return answer
```
