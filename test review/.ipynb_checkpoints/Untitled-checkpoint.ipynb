```python
def solution(acts):
    idx = 0
    max_idx = 0 
    back_stack = []
    forward_stack = []
    answer = [] #act가 실행뒤 
    
    for act in acts:
        # i in range(act[11):
        print("idx", idx, "\n", "back_stack",back_stack,"\n","forward_stack", forward_stack, "\n","act", act)
        if act[0] == 0 : # 페이지 이동
            for i in range(act[1]): #위로 가도 되지만,나중에 횟수카운트 한번에 하기 위해서 
                back_stack.append(idx) #현재 페이지
                forward_stack =[] 
                max_idx += 1
                idx = max_idx
            answer.append(idx)
        if act[0] == -1 : #뒤로가기
            for j in range(act[1]):
                if back_stack : #비어있음 무시
                    forward_stack.append(idx) # 현재 페이지 먼저 앞으로
                    idx = back_stack.pop( )
            answer.append(idx)
        if act[0] == 1 : # 앞으로 가기
            for k in range(act[1]):
                if forward_stack:
                    back_stack.append(idx)
                    idx = forward_stack.pop()
            answer.append(idx)
    return answer 
```

```python
def solution(acts):
    idx = 0
    max_idx = 0 
    back_stack = []
    forward_stack = []
    answer = [] #act가 실행뒤 
    
    for act in acts:
        # i in range(act[1):
        print("idx", idx, "\n", "back_stack",back_stack,"\n","forward_stack", forward_stack, "\n","act", act)
        if act[0] == 0 : 
#         if act[0] == 0 and idx != 0:  # 최초 시행을 제외하고
            back_stack.append(idx)
            back_stack += list(range(max_idx,max_idx+act[1]-1))
            forward_stack =[] 
            max_idx += act[1]
            idx = max_idx
            answer.append(idx)
            
        if act[0] == -1 : #뒤로가기
            forward_stack.append(idx) # 현재 페이지 먼저 앞으로
            forward_stack += back_stack[-act[1]:][::-1]
            back_stack = back_stack[:-act[1]]
            idx = forward_stack.pop()
            answer.append(idx)
            
        if act[0] == 1 : # 앞으로 가기
            back_stack.append(idx)
            back_stack += forward_stack[-act[1]:][::-1]
            forward_stack = forward_stack[:-act[1]]
            idx = back_stack.pop()
            answer.append(idx)
    return answer 
```

```python
def solution(acts):
    idx = 0
    max_idx = 1
    back_stack = []
    forward_stack = []
    answer = [] # act가 실행 후 
    
    first_zero = True  # 첫 번째 0을 표시하기 위한 플래그
    
    for act in acts:
        # Debugging prints
        print("idx", idx, "\n", "back_stack",back_stack,"\n","forward_stack", forward_stack, "\n","act", act)
        
        if act[0] == 0 : 
            if not first_zero:  # 첫 번째 0을 제외하고만 back_stack에 추가
                back_stack.append(idx)
                
            back_stack += list(range(max_idx, max_idx + act[1]))
            forward_stack = [] 
            max_idx += act[1]
            idx = max_idx
            answer.append(idx)
            
            first_zero = False  # 첫 번째 0이 실행되었으므로 플래그를 변경
            
        if act[0] == -1 : # 뒤로가기
            forward_stack.append(idx) # 현재 페이지 먼저 앞으로
            forward_stack += back_stack[-act[1]:][::-1]
            back_stack = back_stack[:-act[1]]
            idx = forward_stack.pop()
            answer.append(idx)
            
        if act[0] == 1 : # 앞으로 가기
            back_stack.append(idx)
            back_stack += forward_stack[-act[1]:][::-1]
            forward_stack = forward_stack[:-act[1]]
            idx = back_stack.pop()
            answer.append(idx)
            
    return answer
```

```python
acts = [[0,4],[-1,3],[0,4],[-1,2],[1,3]]
solution(acts)
```
