7-LED 디스플레이는 7개의 LED를 사용해 0에서 9까지의 숫자를 표시할 수 있는 장치입니다.

- A~ G로 분류되는 LED에 스위치를 연결해 LED를 점등할 수 있습니다.
- 표시하려는 숫자에 따라 점등할 LED를 조합해 숫자를 표시할 수 있습니다
- 위 그림과 정확하게 일치하는 LED 조합으로 숫자를 표시해야 하며, 그 외의 방법은 잘못된 표시 방법입니다.예를 들어, [E, F]을 점등하면 1을 표시한 것으로 인정되지 않습니다.  
각 숫자별로 점등해야 하는 LED를 표로 나타내면 아래와 같습니다.  


LED  
0 : A, B, C, D, E ,F   
1 : B, C  
2 : A, B, D, E, G   
3 : A, B, C, D, G  
4 : B, C, F, G   
5 : A, C, D, F, G   
6 : A, C, D, E, F, G  
7 : A, B, C, F  
8 : A, B, C, D, E, F, G   
9 : A, B, C, D, F, G  

당신은 7-LED 디스플레이를 사용해 정해진 숫자를 순서대로 표시해야 합니다. 이를 위해 각 LED를 최대 한 개의 스위치에 연결할 수 있으며, 표시해야 하는 숫자의 조합에 따라 여러 LED를 같은 스위치에 연결하면 필요한 스위치 의 개수를 줄일 수도 있습니다.  
예를 들어 표시하려는 숫자가 [3, 8, 3, 9 ]라고 가정해 보겠습니다.  
A~ GLED를 7개의 스위치에 하나씩 연결한 뒤, 아래 그림과 같이 스위치를 조작하면 3, 8, 3, 9를 표시할 수 있습니다.
  
당신은 표시하려는 숫자 조합에 따라 필요한 스위치의 최소 개수를 알고 싶습니다.  
표시하려는 숫자 조합들을 담은 1차원 문자열 배열 arr 이 매개변수로 주어집니다. 이때, 각 숫자 조합마다 필요한 스위치의 최소 개수를 arr 에서 주어진 순서대로 1차원 정수 배열에 담아 return 하도록 solution 함수를 완성해 주세요.


제한사항
- 2 <= arr 의 길이 <= 1,000
    - arr 의 원소는 0 이상 9 이하의 숫자만으로 이루어진 길이 10 이하의 문자열입니다.

```python
def solution(arr):
    # LED 조합 정보 딕셔너리
    led_map = {
        '0': 'ABCDEF',
        '1': 'BC',
        '2': 'ABDEG',
        '3': 'ABCDG',
        '4': 'BCFG',
        '5': 'ACDFG',
        '6': 'ACDEFG',
        '7': 'ABC',
        '8': 'ABCDEFG',
        '9': 'ABCDFG'
    }
    
    answer = []
    
    for nums in arr:
        # 숫자 조합에 포함된 모든 LED 조합 합집합
        all_leds = set()
        for num in nums:
            all_leds.update(led_map[num])
        
        # 필요한 최소 스위치 개수는 합집합의 길이
        min_switches = len(all_leds)
        answer.append(min_switches)
    
    return answer
```
