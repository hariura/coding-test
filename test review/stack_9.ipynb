```python
# 예외처리로 해결 최종 코드
def solution(acts):
    idx = 0
    max_idx = 0 
    back_stack = []
    forward_stack = []
    answer = [] #act가 실행뒤 
    is_first_act = True

    for act in acts:
        print("idx", idx, "\n", "back_stack",back_stack,"\n","forward_stack", forward_stack, "\n","act", act)
        if act[0] == 0:
            if is_first_act:
                back_stack += list(range(max_idx + 1, max_idx + act[1]))
                is_first_act = False
            else:
                back_stack.append(idx)
                back_stack += list(range(max_idx + 1, max_idx + act[1]))
                forward_stack = []
            max_idx += act[1]
            idx = max_idx
            answer.append(idx)
            
        if act[0] == -1 : #뒤로가기
            forward_stack.append(idx) # 현재 페이지 먼저 앞으로
            forward_stack += back_stack[-act[1]:][::-1]
            back_stack = back_stack[:-act[1]]
            idx = forward_stack.pop()
            answer.append(idx)
            
        if act[0] == 1 : # 앞으로 가기
            back_stack.append(idx)
            back_stack += forward_stack[-act[1]:][::-1]
            forward_stack = forward_stack[:-act[1]]
            idx = back_stack.pop()
            answer.append(idx)
    return answer 
```

```python
acts = [[0,4],[0,4]]
solution(acts)
```
