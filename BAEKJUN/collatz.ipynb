프린트를 꼭 써야하는줄 알았는데 여긴 또 아니다 ;

```python
def collatz(N, answer):
    if N == 1:
        print(N, end=" ")
        return
    print(N, end=" ")
    if N % 2 == 0:
        return collatz(N // 2, answer + 1)
    else:
        return collatz(N * 3 + 1, answer + 1)

# 입력을 받습니다.
N = int(input().strip())

# Collatz 수열을 출력합니다.
collatz(N, 0)
```
