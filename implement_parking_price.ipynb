```python
from collections import defaultdict
def solution(fees, records):
    answer = []

    # 전처리
    r = defaultdict(list) 
    for record in records:
        time, num, inout = record.split(" ")
        min_time = int(time[0:2])*60 + int(time[3:])
        r[num] +=  [min_time] 

    # 예외처리 
    for k, v in r.items() : 
        if len(v)%2 == 1 :
            r[k] += [23*60+59]
            
    # cost 계산 
    cost = []
    for key in sorted(r.keys()):
        cost_min = 0
        for i in range(len(r[key])//2): 
            cost_min += r[key][2*i+1] - r[key][2*i]
        cost.append(cost_min)

    # 전체 fee 계산 
    for c in cost:
        if c < fees[0] : 
            answer.append(fees[1])
        else:
            m = (c-fees[0])//fees[2] if (c-fees[0])%fees[2] == 0 else (c-fees[0])//fees[2] + 1
            answer.append( fees[1] + (m)*fees[3])
            
    return answer
```
