continue가 점프 하는거였어 인덱스 하나만 넘어감 
pass는 오류 안나게 하는거 

Break 그 for 문 안에서 루프 멈추는거 다음 루프로 넘어감 

```python

import sys 

sys.setrecursionlimit(3000)

def solution(board, moves):
    stack = [0] #공갈스택 
    answer = 0
    
    # move로 선택되는 애 고르기
    for i in moves:
        for m in range(len(board)):
            if board[m][i-1] != 0:
                break 
                
        if board[m][i-1] == 0 : continue 
    
        if stack[-1] == board[m][i-1]:
            stack.pop()
            answer += 2

        else : stack.append(board[m][i-1])
        
        board[m][i-1] = 0 
    return answer
```
