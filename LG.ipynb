여러 시작점에서 BFS를 사용하여 가장 가까운 클래스를 찾고, 같은 거리에 있는 클래스가 여러 개일 경우 가장 적게 방문된 클래스를 선택하되, 방문 횟수도 동일하다면 클래스 번호가 작은 것을 선택하는 문제다.

```python
from collections import deque, defaultdict

def solution(maps, starts):
    answer = []
    dx = [-1,1,0,0]
    dy = [0,0,-1,1]
    class_cord = defaultdict(list)
    class_cnt = defaultdict(int)
    for m in range(len(maps)):#행
        for n in range(len(maps[0])):#렬
            if maps[m][n] != 0 : 
                class_cnt[maps[m][n]] = 0  # -1은 무시하기 
                class_cord[maps[m][n]] += [m,n]
    
    
    for start in starts: 
        queue = deque()
        visited = [[0]*len(maps[0]) for _ in range(len(maps))] # 방문 처리를 거리로 업데이트 하자 
        # 몇 번만에 갔는지 상관없어. 그냥 위치만 중요 
        # 한개가지고만해보자
        # start = starts[0] # 날라가라 부분 
        x,y = start[0] - 1 , start[1] - 1 
        visited[x][y] = 1 
        queue.append([x,y])
        while queue : 
            x,y = queue.popleft()
            for i in range(4):
                nx , ny = x + dx[i], y + dy[i]
                if -1< nx <len(maps) and  -1< ny< len(maps[0]):
                    if visited[nx][ny] == 0 and maps[nx][ny] != -1 : 
                        queue.append([nx , ny])
                        visited[nx][ny] = visited[x][y] + 1 
        
        # print(class_cord)
        save_class_dist = [ ]
        for k,v in class_cord.items():
            save_class_dist.append([k, visited[v[0]][v[1]]])
        ready_append = sorted(save_class_dist, key= lambda x : x[1])
        if ready_append[0][0] == -1 : ready_append.pop(0)
        print(ready_append)
        print(class_cnt)
        ll = [l[0] for l in ready_append if l[1] == ready_append[0][1]]
        # print("cnt ready ", ll) # ll에 들어있는 클래스중 cnt가 작은 애 
        care_class = 0
        care_cnt = class_cnt[ll[0]]
        for i in ll[::-1]:
            if class_cnt[i] <= care_cnt:
                care_cnt = min(care_cnt,class_cnt[i])
                care_class = i

        # aa = ready_append.pop(0)[0]
        answer.append(care_class)
        class_cnt[care_class] += 1 

    
    
    return answer
```
