**횟수를 카운트 해야하는게 중요**
- 1차원 DP로만 업데이트 생각 했었음
- 2차원 생각했을 때, 문제를 제대로 파악하지 못함 목표 점수를 맞추는 테이블을 만드려고 함 ( 이전에 풀어봤으니)
- 원하는 목표가 뭔지 제대로 생각
    - 테이블의 모양과 숫자 등을 잘 고려했었어야 함

```python
def solution(N, number):
    # N : 숫자 , number : 목표 타겟 
    dp = [set() for i in range(9)]
    for i in range(1,9):
        case = dp[i]
        case.add(int(str(N)*i))
        
        for j in range(1,i):
            for k in dp[j]:
                for l in dp[i-j]:
                    case.add(k+l)
                    case.add(k-l)
                    case.add(k*l)
                    if l != 0 and k != 0 : case.add(k//l)
        print(case)
                    
        if number in case : return i
    
    return -1 
```
