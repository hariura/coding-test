```python
count = 0
for i in range(38):
    if "3" in "%s" % i   :
        count +=1
        print(i)
    if  "7" in "%s" % i : 
        count +=1
        print(i)
```

```python
# 아래껀 ok 
count = 0
for i in range(1,38):
    for ch in str(i): # 이렇게 하면 자리수 하나씩 들어가잖아. 
        if ch == "3" or ch == "7":
            count +=1
```
