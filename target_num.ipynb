- 내가 과연 DFS를 코드로 구현할 수 있을까 ? 
- 코드를 적고 나서도 내혼자 스스로 할 수 있다는 생각이 전혀 들지 않아 
- 거의 무슨 기도메타. 

```python
def dfs(numbers, step, total, target):
    if step == len(numbers):
        if total == target : 
            return 1
        else : return 0
    
    return dfs(numbers, step+1, total + numbers[step+1], target) + dfs(numbers, step+1, total - numbers[step+1], target)

def solution(numbers, target):
    return dfs(numbers, 0,0, target)
```
