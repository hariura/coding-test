```python
#지금의 이 dfs는 그냥 for문이야 순서를 바꾸는게 전혀 안들어가있어 
# 어디까지 들어가는지 여기서 나와야하고
# 아래 solution에선 몇번 들어가는지(x,y방향)이 나와야해 
# 최종적으로 세고 싶은건 던전에 몇번 갔냐
# 보통의 dfs는 다음 지정의 노드가 들어가. 
def dfs(dungeons, visited, pirodo, cnt): 
    answer_max = cnt
    for i in range(len(dungeons)):
        if not visited[i] and pirodo >= dungeons[i][0] and pirodo >= dungeons[i][1]:
            #문제에 두번째 조건을 명시하였어서 나타낼 필요 없었음 
            visited[i] = 1
            answer_max = max(answer_max, dfs(dungeons, visited , pirodo - dungeons[i][1], cnt+1 ))
            visited[i] = 0 
    return answer_max
                   

def solution(k, dungeons):
    visited = [0] * len(dungeons)
    return dfs(dungeons, visited, k, 0)
```
