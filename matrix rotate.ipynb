```python
def rotate_90_degree(matrix):
    n = len(matrix)
    m = len(matrix[0])
    
    # Step 1: Transpose the matrix
    transposed = []
    for i in range(m):
        transposed_row = []
        for j in range(n):
            transposed_row.append(matrix[j][i])
        transposed.append(transposed_row)

    # Step 2: Reverse each row
    for row in transposed:
        row.reverse()
        
    return transposed

matrix = [
    [1, 2, 3,4],
    [4, 5, 6,5],
    [7, 8, 9,6]
]

rotated_matrix = rotate_90_degree(matrix)
```
