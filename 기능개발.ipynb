```python
from collections import deque

def solution(progresses, speeds):
    answer = []
    queue = deque(progresses)
    
    while queue:
        for i in range(len(queue)): # 한번 더해지기 
            queue[i] += speeds[i]
        a = 0
        while queue and queue[0] >= 100 : # 100넘는게 있으면  
            queue.popleft() ; speeds.pop(0) #원본 건들기 때문에 큐 사이즈가 작아져서 아웃오브레인지 
            a += 1 
        if a != 0 : answer.append(a)
    
    return answer
```
