```python
lst = [1, 2, 3, 4, 5, 6, 7, 8, 9]

# 리스트의 길이의 절반(내림)을 계산
n = len(lst) // 2

# 리스트의 양 끝에서 중앙으로 진행하며 쌍 생성
pairs = [(lst[i], lst[-i-1]) for i in range(n + 1)]

print(pairs)
```
