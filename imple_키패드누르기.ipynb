LV1 논리 처리가 조금 어렵긴 해도 문제는 풀었음   
강의 기준으로 하니까 되게 멋있는 코드가 나오는 것 같음  
왜냐 if else 구문이 하나가 줄어듬 and 5줄 이상 줄어듬   
아이디어는 하나를 선택하고 그게 아니면 다른걸 선택하는것   

```python
def solution(numbers, hand):
    answer = ''
    l_temp, r_temp = [3,0] , [3,2]
    for n in numbers:
        if n in [1,4,7]:
            answer += "L"
            l_temp = [(n - 1) /3, 0 ]     
        elif n in [3,6,9]:
            answer += "R"
            r_temp = [(n - 3) /3, 2 ]
            
        elif n in [2,5,8,0]:
            if n == 0 : target = [3,1]
            else : target = [(n-2)/3, 1] 
            
            if abs(r_temp[1] - target[1]) + abs(r_temp[0] - target[0]) > abs(l_temp[1] - target[1]) + abs(l_temp[0] - target[0]) : 
                answer += "L" 
                l_temp = target
            elif abs(r_temp[1] - target[1]) + abs(r_temp[0] - target[0]) < abs(l_temp[1] - target[1]) + abs(l_temp[0] - target[0]):
                answer += "R"
                r_temp = target
            else :
                if hand == "left":
                    answer += "L" 
                    l_temp = target
                else : 
                    answer += "R"
                    r_temp = target
        # print(l_temp, r_temp)
                
    return answer
```

```python
def dist(target, pos):
    return abs(target[0]- pos[0]) + abs(target[1] - pos[1])

def solution(numbers, hand):
    pattern = {"L" : (1,4,7) , 'R' : (3,6,9)}
    pos = {'L': [3,0], 'R' : [3,2]}
    result = []
    
    def press(which , coord):
        result.append(which) 
        pos[which] = coord
    
    for number in numbers:
        choose = 'L'
        target = [(number - 1)//3, 0]
        
        if number in pattern["L"]: pass
        elif number in pattern["R"] : choose = 'R'
        else : 
            target = [3 if number == 0 else (number - 1)//3 , 1]
            left = dist(target, pos['L'])
            right = dist(target, pos['R'])
            
            if right < left : choose = 'R'
            elif right == left and hand == 'right' : choose = 'R'
        
        press(choose, target)
            

    return ''.join(result)
```
