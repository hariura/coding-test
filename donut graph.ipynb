```python
from collections import defaultdict

def solution(edges):
    node_info = defaultdict(lambda: {'give': 0, 'receive': 0})
    max_num = 0

    for give_num, receive_num in edges:
        max_num = max(max_num, give_num, receive_num)

        node_info[give_num]['give'] += 1
        node_info[receive_num]['receive'] += 1

    answer = [0, 0, 0, 0]

    for i in range(1, max_num + 1):
        give_count = node_info[i]['give']
        receive_count = node_info[i]['receive']

        if receive_count == 0 and give_count >= 2:
            answer[0] = i
        elif give_count == 0:
            answer[2] += 1
        elif receive_count >= 2 and give_count >= 2:
            answer[3] += 1

    answer[1] = node_info[answer[0]]['give'] - answer[2] - answer[3]

    return answer
```
