```python
def solution(plans):
    answer = []
    stack = []
    
    for post_name, post_start_time, post_play_time in sorted(plans, key = lambda p : p[1]):
        post_start_time, post_play_time = convert_time(post_start_time), int(post_play_time)
        
        if not stack:
            stack.append([post_name, post_start_time, post_play_time])
            continue
        
        _, curr_start_time, _ =stack[-1]
        working_time = post_start_time - curr_start_time
        while 0 < working_time and stack : 
            curr_name, curr_start_time, curr_play_time = stack.pop()
            
            if working_time < curr_play_time:
                curr_remain_time = curr_play_time - working_time
                stack.append([curr_name, curr_start_time, curr_remain_time])
                break
            
            working_time -= curr_play_time
            answer.append(curr_name)
            
        stack.append([post_name, post_start_time, post_play_time])
        
    while stack:
            stack_name, _, _ = stack.pop()
            answer.append(stack_name)
            
        
    return answer
    
    


def convert_time(time):
    hh, mm = map(int, time.split(":"))
    
    return hh*60 + mm
    
    
    
    
    
    
    
    
    
    

    
```
