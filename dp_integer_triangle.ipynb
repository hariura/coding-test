```python
#내가 푼것에 대한 문제점 찾아라 빨리
def solution(triangle):
    # 삼각형의 층수
    length = len(triangle)

    # 동적 프로그래밍을 위한 초기 설정: 맨 아래층으로 시작
    dp = triangle[-1]

    # 맨 아래층 바로 위부터 시작하여 맨 위층까지 이동
    for l in range(length - 2, -1, -1):
        for i in range(len(triangle[l])):
            # 현재 위치에서 아래층의 두 경로 중 더 큰 값 선택
            dp[i] = triangle[l][i] + max(dp[i], dp[i + 1])

    # 최종 결과 반환 (맨 위층의 값)
    return dp[0]
```
